0. Cargar librerias


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

1. Cargar base de datos de trabajo

```python
nxl= '/content/Base_de_Datos_FINTECH_LDA_v2.xlsx'
XDB= pd.read_excel(nxl,sheet_name=0)
display (XDB)
Freq= XDB.iloc[:,3] #para seleccionar la colunma "freq" de la tablita
Sev=XDB.iloc[:,4] #lo mismo pero para severidad
LDA= Freq*Sev
LDA
```

```python
plt.figure()
sns.kdeplot(LDA)
plt.grid()
plt.show()
```

2. Se procede con la creación de los concentradores de información.

```python
import random
random.seed(42)
XC=np.array(random.choices (LDA,k=5))
XC= np.sort(XC) #ORGANIZA DE MENOR A MAYOR
print("Los concentradores de información son:",XC)
```

3. Se clasifican los datos de acuerdo con los concentradores de información

```python
d= np.abs(XC-LDA.iloc[0,]) #La distancia de un dato a un concentrador
nc= np.argmin(d)
print("El dato pertenece a la categoria:",nc)
# aca nos dice ell dato seleccionado en que categoria esta (a cual se parece más)
```

```python
box0= [];box1=[];box2= [];box3=[];box4= []
Boxes=[box0,box1,box2,box3,box4]
for k in range(len(LDA)):
    d= np.abs(XC-LDA.iloc[k,])/XC
    #Acá se divide para que de numeros mas pequeños (entre 0 y 1)
    nc= np.argmin(d) # Numero de la caja en donde voy a meter el dato
    Boxes[nc].append(LDA.iloc[k,])

Boxes[0]  #CLICK DERECHO EJECUTAR SELECCION
Boxes[4] # perdidads catastroficas
```

```python
# CUANTOS DATOS HAY EN CADA CAJA
print(len(Boxes[0]),len(Boxes[1]),len(Boxes[2]),len(Boxes[3]),len(Boxes[4]))
# DETERMINAR EL PORCENTAJE DE DATOS POR CAJA


pm=[] #Vector dde porcentajes
for j in range (5):
  pd=len(Boxes[j])/len(LDA)
  pm.append(pd)
print("El porcentaje de los datos es:",pm)
pm= (np.array(pm)*2000).astype(int)
print ("EL porcentaje de datos que se usan es:",pm) # cuantos datos saco de cada caja si necesito mil datos
```

4. Sacamos datos por caja y creamos una LDA 2

```python
import pandas as pd # Re-import pandas to ensure 'pd' is correctly defined

LDA2=[] #Aqui van a quedar los 2000 datos
# random.choices(Boxes[0],k=379)

for j in range(5):
  LDA2. extend(random.choices(Boxes[j],k=pm[j]))
LDA2=np.array (LDA2)

print(LDA2.shape)

# se comprueban las metricas estadisticas
uo=np.mean(LDA) #datos observados son los originales
print("La media de los datos observados es:",uo)
ue=np.mean(LDA2) #datos externos, los que seleccionamos
print("La media de los datos externos es:",ue)

vo=np.var(LDA)
print("La varianza de los datos observados es:",vo)
ve=np.var(LDA2)
print("La varianza de los datos externos es:",ve)

sdo=np.std(LDA)
print("La desviación estándar de los datos observados es:",sdo)
sde=np.std(LDA2)
print("La desviación estándar de los datos externos es:",sde)

skeo=pd.Series(LDA).skew() # Calculate skewness for LDA
print("La asimetría de los datos observados es:",skeo)
skee=pd.Series(LDA2).skew() # Calculate skewness for LDA2
print("La asimetría de los datos externos es:",skee)

ko=pd.Series(LDA).kurt() # Ensure LDA is treated as a Series for kurtosis
print("La curtosis de los datos observados es:",ko)
ke=pd.Series(LDA2).kurt()
print("La curtosis de los datos externos es:",ke)

du=(uo-ue)/uo
print("El error relativo porcentual es:",du*100)

plt.figure()
sns.kdeplot(LDA, color='red', label='Datos Observados')
sns.kdeplot(LDA2, color='blue', label='Datos Externos')
plt.grid()
plt.show()
```

El error relativo porcentual del 1.37% demuestra que el comportamiento promedio del riesgo interno está fuertemente calibrado con la referencia externa de la industria.

Ambas fuentes presentan una asimetría marcadamente positiva (>3.0) y una curtosis leptocúrtica extrema (>16.0). Esto evidencia que los eventos extremos (pérdidas severas de baja frecuencia) dominan la estructura de riesgo en ambos conjuntos.

5. Metodo montecarlo

```python
print('--- Calculando confiabilidad para LDA (Datos Observados) ---')
# Ordenar el array LDA
LDA_sorted = np.sort(LDA)

# Calcular el índice para el percentil 99.9%
# Multiplicamos por la longitud y restamos 1 para obtener el índice basado en 0
index_99_9_LDA = int(len(LDA_sorted) * 0.999) - 1
# Asegurarse de que el índice no exceda el tamaño del array
if index_99_9_LDA < 0: index_99_9_LDA = 0
if index_99_9_LDA >= len(LDA_sorted): index_99_9_LDA = len(LDA_sorted) - 1

# Obtener el valor en ese percentil
conf_99_9_LDA = LDA_sorted[index_99_9_LDA]
print(f"Confiabilidad del 99.9% para LDA: {conf_99_9_LDA:.4f}")

# Calcular el índice para el percentil 99.99%
index_99_99_LDA = int(len(LDA_sorted) * 0.9999) - 1
if index_99_99_LDA < 0: index_99_99_LDA = 0
if index_99_99_LDA >= len(LDA_sorted): index_99_99_LDA = len(LDA_sorted) - 1

conf_99_99_LDA = LDA_sorted[index_99_99_LDA]
print(f"Confiabilidad del 99.99% para LDA: {conf_99_99_LDA:.4f}")

print('\n--- Calculando confiabilidad para LDA2 (Datos Externos) ---')
# Ordenar el array LDA2
LDA2_sorted = np.sort(LDA2)

# Calcular el índice para el percentil 99.9%
index_99_9_LDA2 = int(len(LDA2_sorted) * 0.999) - 1
if index_99_9_LDA2 < 0: index_99_9_LDA2 = 0
if index_99_9_LDA2 >= len(LDA2_sorted): index_99_9_LDA2 = len(LDA2_sorted) - 1

conf_99_9_LDA2 = LDA2_sorted[index_99_9_LDA2]
print(f"Confiabilidad del 99.9% para LDA2: {conf_99_9_LDA2:.4f}")

# Calcular el índice para el percentil 99.99%
index_99_99_LDA2 = int(len(LDA2_sorted) * 0.9999) - 1
if index_99_99_LDA2 < 0: index_99_99_LDA2 = 0
if index_99_99_LDA2 >= len(LDA2_sorted): index_99_99_LDA2 = len(LDA2_sorted) - 1

conf_99_99_LDA2 = LDA2_sorted[index_99_99_LDA2]
print(f"Confiabilidad del 99.99% para LDA2: {conf_99_99_LDA2:.4f}")
```

La exigencia de capital base se sitúa en 2.28 MUSD con datos propios frente a 2.15
MUSD con datos externos. La escasa diferencia (5.83%) ratifica la validez y robustez del modelo de pérdida agregada.

El percentil 99.9% interno es matemáticamente al percentil 99.99% externo (2,280.01KUSD), lo que permite usar el percentil 99.99% externo como piso mínimo de protección.

En eventos extremos de muy baja frecuencia (como caídas prolongadas del core o ciberataques masivos), la severidad interna escala hasta 2.87 MUSD, superando en un 20.48% al referente externo. Esto se explica por la densidad de la cola pesada en la operación real.
